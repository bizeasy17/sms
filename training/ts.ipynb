```python
df = pro.stk_factor(ts_code='000001.SZ')
df.head()
```

```python
import matplotlib.pyplot as plt
from scipy.signal import find_peaks
from datetime import datetime, timedelta

# 计算过去5年的日期
end_date = df["trade_date"].iloc[0]
start_date = (datetime.strptime(end_date, "%Y%m%d") - timedelta(days=5*365)).strftime("%Y%m%d")

# 只保留过去5年的数据
df_5y = df[df["trade_date"] >= start_date].reset_index(drop=True)

# 找到峰值和谷值
peaks, _ = find_peaks(df_5y["close_qfq"], distance=20)
valleys, _ = find_peaks(-df_5y["close_qfq"], distance=20)
df = df_5y  # 后续绘图使用df_5y

plt.figure(figsize=(16, 6))
plt.plot(df["close_qfq"], label="Close QFQ", color='blue')
plt.scatter(peaks, df["close_qfq"].iloc[peaks], color='red', label='Peaks', marker='^')
plt.scatter(valleys, df["close_qfq"].iloc[valleys], color='green', label='Valleys', marker='v')
plt.legend()
plt.title("Close QFQ with Peaks and Valleys")
plt.xlabel("Index")
plt.ylabel("Price")
plt.show()
```
